### Histogram: Distribution of Data

A histogram is used to visualize the distribution of a single numerical variable. It divides the data into bins and shows the frequency of data points falling into each bin.

```python
import numpy as np

# Generate some sample numerical data for the histogram
data_for_histogram = np.random.randn(1000) # 1000 random numbers from a standard normal distribution

plt.figure(figsize=(8, 5))
plt.hist(data_for_histogram, bins=30, color='skyblue', edgecolor='black')
plt.title('Distribution of Sample Numerical Data (Histogram)')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

### Bar Chart: Comparing Categorical Data

A bar chart is excellent for comparing values across different categories. Let's create one to show hypothetical sales per region.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Sample data for a bar chart
regions = ['North', 'South', 'East', 'West']
sales = [150, 220, 180, 100]

plt.figure(figsize=(8, 5))
sns.barplot(x=regions, y=sales, hue=regions, palette='viridis', legend=False)
plt.title('Hypothetical Sales by Region')
plt.xlabel('Region')
plt.ylabel('Sales (in units)')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

### 3D Scatter Plot: Visualizing Three Variables

A 3D scatter plot allows you to visualize the relationship between three numerical variables (X, Y, Z) in a three-dimensional space.

```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

# Generate some sample 3D data
n = 100
x_data = np.random.rand(n)
y_data = np.random.rand(n)
z_data = np.random.rand(n)

# Create a new figure and add a 3D subplot
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

# Plot the 3D scatter points
ax.scatter(x_data, y_data, z_data, c='blue', marker='o')

# Set labels and title
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')
ax.set_title('Sample 3D Scatter Plot')

plt.show()
```

### 3D Wave Plot: Visualizing a 3D Sine Wave Surface

This plot shows a 3D representation of a sine wave, typically seen as a surface oscillating in three dimensions.

```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

# Create a meshgrid for x and y values
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)

# Calculate z values for a 3D sine wave
Z = np.sin(np.sqrt(X**2 + Y**2))

# Create the 3D plot
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Plot the surface
ax.plot_surface(X, Y, Z, cmap='viridis')

# Set labels and title
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')
ax.set_title('3D Sine Wave Surface Plot')

plt.show()
```

### 3D Bar Plot: Visualizing Categorical Data in 3D

This plot extends the concept of a 2D bar chart into three dimensions, allowing for the representation of data with three categorical or two categorical and one numerical axis.

```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

# Sample data for 3D bar plot
xpos = np.arange(4)
ypos = np.arange(5)
x, y = np.meshgrid(xpos, ypos)
x = x.flatten()
y = y.flatten()
z = np.zeros_like(x)

dx = 0.8 * np.ones_like(x)
dy = 0.8 * np.ones_like(y)
dz = np.random.rand(len(x)) * 10 # Heights of the bars

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

ax.bar3d(x, y, z, dx, dy, dz, color='cyan', alpha=0.8)

ax.set_xlabel('Category X')
ax.set_ylabel('Category Y')
ax.set_zlabel('Value Z')
ax.set_title('Sample 3D Bar Plot')

plt.show()
```
